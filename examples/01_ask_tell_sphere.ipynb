# Swarm Seek: ask/tell on Sphere

Demonstrate the public ask/tell loop on the Sphere objective.

## Verification objective

- Seeded run completes without error
- Best fitness is below `1e-2`

```python
from __future__ import annotations

import numpy as np

from swarm_seek import ABC, ContinuousSpace
from swarm_seek.benchmarks import sphere

SEED = 0
MAX_EVALS = 5_000
PASS_THRESHOLD = 1e-2

space = ContinuousSpace([(-5.0, 5.0)] * 5)
colony = ABC(
    space,
    variant="original",
    pop_size=20,
    limit=100,
    max_evals=MAX_EVALS,
    seed=SEED,
)

while not colony.converged:
    candidates = colony.ask()
    if candidates.size == 0:
        colony.tell([])
        continue
    fitnesses = np.asarray(sphere(candidates), dtype=np.float64)
    colony.tell(fitnesses)

best = colony.best
print(f"best fitness={best.fitness:.6e}, n_evals={colony.n_evals}, n_iters={colony.n_iters}")
assert best.fitness < PASS_THRESHOLD, (
    f"Pass criterion failed: best fitness {best.fitness!r} >= {PASS_THRESHOLD}"
)
print("PASS: ask/tell Sphere below threshold")
```
