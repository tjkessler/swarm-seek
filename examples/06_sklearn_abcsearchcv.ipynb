# Swarm Seek: ABCSearchCV

Tiny Ridge hyperparameter search via `ABCSearchCV`.

Requires `pip install 'swarm-seek[sklearn]'`.


```python
from __future__ import annotations

import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import Ridge

from swarm_seek.integrations import ABCSearchCV

x, y = make_regression(n_samples=40, n_features=3, noise=0.1, random_state=0)
search = ABCSearchCV(
    Ridge(),
    param_distributions={"alpha": [0.1, 1.0, 10.0]},
    n_iter=8,
    pop_size=4,
    cv=3,
    seed=0,
)
search.fit(x, y)
print("best_params_", search.best_params_, "best_score_", search.best_score_)
assert np.isfinite(search.best_score_)
print("PASS")
```
