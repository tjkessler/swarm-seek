# Swarm Seek: ABCSampler

Tiny Optuna study using `ABCSampler`.

Requires `pip install 'swarm-seek[optuna]'`.


```python
from __future__ import annotations

import optuna

from swarm_seek.integrations import ABCSampler

def objective(trial):
    x = trial.suggest_float("x", -2.0, 2.0)
    y = trial.suggest_float("y", -2.0, 2.0)
    return x * x + y * y

sampler = ABCSampler(pop_size=4, seed=0)
study = optuna.create_study(direction="minimize", sampler=sampler)
study.optimize(objective, n_trials=20)
print("best_value", study.best_value)
assert study.best_value < 2.0
print("PASS")
```
