# Swarm Seek: variant comparison

Compare Original ABC, GABC, qABC, and MABC on Sphere under matched budgets.

## Verification objective

- All four variants complete without exception
- Table of best fitness (and evaluation counts) is produced

```python
from __future__ import annotations

import numpy as np

from swarm_seek import ABC, ContinuousSpace
from swarm_seek.benchmarks import sphere

SEED = 1
MAX_EVALS = 4_000
DIM = 5
VARIANTS = ("original", "gabc", "qabc", "mabc")

space = ContinuousSpace([(-5.0, 5.0)] * DIM)
rows: list[tuple[str, float, int]] = []

for name in VARIANTS:
    colony = ABC(
        space,
        variant=name,
        pop_size=16,
        limit=80,
        max_evals=MAX_EVALS,
        seed=SEED,
    )
    best = colony.minimize(sphere)
    rows.append((name, float(best.fitness), int(colony.n_evals)))

print(f"{'variant':<10} {'best_fitness':>14} {'n_evals':>8}")
print("-" * 36)
for name, fitness, n_evals in rows:
    print(f"{name:<10} {fitness:14.6e} {n_evals:8d}")

assert len(rows) == 4
assert all(np.isfinite(f) for _, f, _ in rows)
assert all(e > 0 for _, _, e in rows)
print("PASS: all four variants completed")
```
