```python
from typing import List

from abc_ml import Colony
from abc_ml.variables import Integer


def objective_function(int_values: List[int]) -> int:
    return sum(int_values)


variables = [Integer(0, 50) for _ in range(10)]

colony = Colony(
    n_employers=64,
    variables=variables,
    objective_fn=objective_function
)
colony.initialize()
```

```python
N_ITER = 512

average = []
best = []
for _ in range(N_ITER):
    best.append(colony.best_fitness)
    average.append(colony.average_fitness)
    colony.search()
```

```python
from matplotlib import pyplot as plt


plt.clf()
plt.plot([i for i in range(N_ITER)], best, color="blue", label="Best fitness")
plt.plot([i for i in range(N_ITER)], average, color="red", label="Average fitness")
plt.xlabel("Search iteration")
plt.ylabel("Fitness")
plt.legend(loc="lower right")
plt.show()
```
