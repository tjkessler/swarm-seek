# Swarm Seek: literature-oracle smoke

Short smoke of the MABC literature-oracle helper (Akay & Karaboga, 2012, Table 2 Sphere).

## Verification objective

- Mean best over a few seeded runs is within the fixture `abs_tol`
- Uses the package oracle helper (not a reimplemented protocol)

```python
from __future__ import annotations

from swarm_seek.benchmarks import (
    AKAY_KARABOGA_2012_T2_SPHERE,
    mean_best_over_runs,
)

oracle = AKAY_KARABOGA_2012_T2_SPHERE
# Paper used 30 runs; smoke uses 2 with the same per-run protocol.
N_RUNS = 2
mean_best = mean_best_over_runs(oracle, n_runs=N_RUNS, seed0=0)
print(
    f"oracle={oracle.id}\n"
    f"mean_best={mean_best:.6e} (n_runs={N_RUNS})\n"
    f"expected~{oracle.expected:.6e}, abs_tol={oracle.abs_tol}\n"
    f"locator={oracle.locator}"
)
assert mean_best <= oracle.abs_tol, (
    f"Literature smoke failed: mean_best={mean_best!r} > abs_tol={oracle.abs_tol}"
)
print("PASS: literature smoke within oracle tolerance")
```
